### **Question 26**
(Visualization) Dataset mein 'Certificate' wise average 'Gross' kamai ka ek Bar Plot banayein.

```python
import matplotlib.pyplot as plt

# Clean Gross column (if needed)
data = data.copy()

data["Gross"] = (
    data["Gross"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .astype(float)
)

# Remove missing values
data = data.dropna(subset=["Certificate", "Gross"])

# Group by Certificate and calculate average gross
avg_gross = data.groupby("Certificate")["Gross"].mean().sort_values(ascending=False)

# Plot Bar Chart
plt.figure(figsize=(8,5))
avg_gross.plot(kind="bar")

plt.title("Average Gross by Certificate")
plt.xlabel("Certificate")
plt.ylabel("Average Gross Revenue")

plt.show()
```

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Clean Gross column
data["Gross"] = pd.to_numeric(
    data["Gross"].astype(str).str.replace(",", "", regex=False),
    errors="coerce"
)

# Select only Drama and Action movies
genre_data = data[data["Genre"].str.contains("Drama|Action", na=False)].copy()

# Create Genre column for comparison
genre_data["Genre_Type"] = genre_data["Genre"].apply(
    lambda x: "Drama" if "Drama" in x else "Action"
)

# Remove missing values
genre_data = genre_data.dropna(subset=["Gross"])

# Box Plot
plt.figure(figsize=(8,5))

sns.boxplot(
    x="Genre_Type",
    y="Gross",
    data=genre_data
)

plt.title("Gross Revenue Comparison: Drama vs Action Movies")
plt.xlabel("Genre")
plt.ylabel("Gross Revenue")

plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Top 5 highest rated movies
top5_movies = data.sort_values(
    by="IMDB_Rating",
    ascending=False
).head(5)

# Convert IMDB Rating to 100 scale
top5_movies["IMDB_Rating_Scaled"] = top5_movies["IMDB_Rating"] * 10

# Remove missing Meta_score values
top5_movies = top5_movies.dropna(subset=["Meta_score"])

# Movie names
movies = top5_movies["Series_Title"]

x = np.arange(len(movies))
width = 0.35

# Create bar chart
plt.figure(figsize=(10,5))

plt.bar(
    x - width/2,
    top5_movies["Meta_score"],
    width,
    label="Meta Score"
)

plt.bar(
    x + width/2,
    top5_movies["IMDB_Rating_Scaled"],
    width,
    label="IMDB Rating (Scaled)"
)

plt.xticks(x, movies, rotation=45, ha="right")

plt.xlabel("Movies")
plt.ylabel("Score (Out of 100)")
plt.title("Top 5 Highest Rated Movies: Meta Score vs IMDB Rating")

plt.legend()
plt.tight_layout()

plt.show()
```
